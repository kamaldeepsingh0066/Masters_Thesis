```python
import requests
from bs4 import BeautifulSoup
import os

base_url = "https://opendata.dwd.de/climate_environment/CDC/observations_germany/climate/hourly/air_temperature/historical/"

response = requests.get(base_url)
soup = BeautifulSoup(response.text, "html.parser")

links = [a.get("href") for a in soup.find_all("a") if a.get("href").endswith((".zip",".pdf",".txt"))]
os.makedirs("Document_Corpus", exist_ok=True)
os.makedirs("Document_Corpus/dwd_temp", exist_ok=True)
os.makedirs("Document_Corpus/dwd_temp/historical_zip", exist_ok=True)
for link in links:
    file_url = base_url + link
    file_path = os.path.join("Document_Corpus/dwd_temp/historical_zip", link)
    
    print("Downloading:", link)
    r = requests.get(file_url)
    with open(file_path, "wb") as f:
        f.write(r.content)
```

# Extrating Solar Zip files

```python
#Extract solar data files into extracted folder
import os
import zipfile
from pathlib import Path
# Folder containing all zip files
zip_folder = Path("Document_Corpus/dwd_solar/zip")

# Output folder for extracted produkt files
output_folder = Path("Document_Corpus/dwd_solar/extracted")
output_folder.mkdir(parents=True, exist_ok=True)

print("ZIP folder:", zip_folder.resolve())
print("Output folder:", output_folder.resolve())



# Check how many ZIP files are available
zip_files = list(zip_folder.glob("*.zip"))
print(f"Total ZIP files found: {len(zip_files)}")

# Show first few zip names
for z in zip_files[:10]:
    print(z.name)

# Extract only files starting with produkt_st_stunde
extracted_files = []
errors = []

for zip_path in zip_files:
    try:
        with zipfile.ZipFile(zip_path, 'r') as z:
            matched = False
            
            for file_name in z.namelist():
                base_name = os.path.basename(file_name)
                
                if base_name.startswith("produkt_st_stunde"):
                    
                    
                    matched = True
                    
                    # Add zip filename prefix to avoid overwriting files
                    #new_name = f"{zip_path.stem}_{base_name}"
                    new_name = f"{base_name}"
                    target_path = output_folder / new_name

                    with z.open(file_name) as source, open(target_path, "wb") as target:
                        target.write(source.read())

                    extracted_files.append(target_path.name)
                    print(f"Extracted: {target_path.name}")
            
            if not matched:
                print(f"No produkt_st_stunde file found in: {zip_path.name}")

    except Exception as e:
        errors.append((zip_path.name, str(e)))
        print(f"Error with {zip_path.name}: {e}")



# Summary
print("="*50)
print("EXTRACTION SUMMARY")
print("="*50)
print(f"ZIP files scanned: {len(zip_files)}")
print(f"Files extracted: {len(extracted_files)}")
print(f"Errors: {len(errors)}")
```

# Merging all solar text files as csv


```python
# Merging all solar text files as csv
import pandas as pd
from pathlib import Path
# Folder containing extracted DWD text files
input_folder = Path("Document_Corpus/dwd_solar/extracted")

# Find all .txt files
txt_files = list(input_folder.glob("*.txt"))

print(f"Total TXT files found: {len(txt_files)}")

# Preview first few files
for f in txt_files[:5]:
    print(f.name)

all_data = []
errors = []

for file_path in txt_files:
    try:
        print(f"Reading: {file_path.name}")
        
        # Read semicolon-separated DWD file
        df = pd.read_csv(
            file_path,
            sep=';',
            comment='#',      # ignore metadata lines if any
            engine='python'
        )
        
        # Optional: add source file name for traceability
        # df["source_file"] = file_path.name
        
        all_data.append(df)
        
    except Exception as e:
        errors.append((file_path.name, str(e)))
        print(f"Error reading {file_path.name}: {e}")




merged_df = pd.concat(all_data, ignore_index=True)
# Remove extra spaces from column names
merged_df.columns = merged_df.columns.str.strip()

print(merged_df.columns.tolist())

print("Merged shape:", merged_df.shape)



# Replace common DWD missing values
merged_df.replace(-999, pd.NA, inplace=True)
merged_df.replace(-999.0, pd.NA, inplace=True)




output_file = Path("Document_Corpus/dwd_solar/final_merged_dwd_solar.csv")

merged_df.to_csv(output_file, index=False)

print(f"Final merged CSV saved at: {output_file.resolve()}")



print("="*50)
print("MERGE SUMMARY")
print("="*50)
print(f"TXT files processed: {len(txt_files)}")
print(f"Rows in final dataset: {len(merged_df)}")
print(f"Columns in final dataset: {len(merged_df.columns)}")
print(f"Errors: {len(errors)}")
merged_df.sample(10)
```

# Power Consumption Data merging

```python
import pandas as pd
from pathlib import Path

# Folder containing CSV files
input_folder = Path("Document_Corpus/power_consumption/seperated")

# Find all CSV files
csv_files = list(input_folder.glob("*.csv"))

print(f"Total CSV files found: {len(csv_files)}")

# Preview first few file names
for f in csv_files[:10]:
    print(f.name)

all_data = []
errors = []

for file_path in csv_files:
    try:
        print(f"Reading: {file_path.name}")
        
        df = pd.read_csv(file_path, sep=';',  engine='python')
        
        # Optional: keep source file name
        df["source_file"] = file_path.name
        
        all_data.append(df)
        
    except Exception as e:
        errors.append((file_path.name, str(e)))
        print(f"Error reading {file_path.name}: {e}")


merged_df = pd.concat(all_data, ignore_index=True)

print("Merged shape:", merged_df.shape)
merged_df.head()

merged_df.columns = merged_df.columns.str.strip()
print(merged_df.columns.tolist())

output_file = Path("Document_Corpus/power_consumption/final_merged_power_consumption.csv")

merged_df.to_csv(output_file, index=False)

print(f"Final merged CSV saved at: {output_file.resolve()}")


print("="*50)
print("MERGE SUMMARY")
print("="*50)
print(f"CSV files processed: {len(csv_files)}")
print(f"Rows in final dataset: {len(merged_df)}")
print(f"Columns in final dataset: {len(merged_df.columns)}")
print(f"Errors: {len(errors)}")
```

```python
# Creating a File Structure of Data Corpus
import os
import pandas as pd

# Base folder (your current main folder)
base_folder = os.getcwd()   # assumes script runs inside Document_Corpus

# Output files
tree_output_file = os.path.join(base_folder, "dwd_file_structure.txt")
inventory_output_file = os.path.join(base_folder, "dwd_file_inventory.csv")

# Store inventory rows
inventory = []

def generate_tree(folder, prefix=""):
    """
    Recursively build folder tree as text.
    """
    entries = sorted(os.listdir(folder))
    tree_lines = []

    for i, entry in enumerate(entries):
        path = os.path.join(folder, entry)
        connector = "└── " if i == len(entries) - 1 else "├── "
        tree_lines.append(prefix + connector + entry)

        # Save file/folder info for CSV inventory
        inventory.append({
            "name": entry,
            "full_path": path,
            "type": "folder" if os.path.isdir(path) else "file",
            "extension": os.path.splitext(entry)[1].lower() if os.path.isfile(path) else "",
            "size_bytes": os.path.getsize(path) if os.path.isfile(path) else None
        })

        if os.path.isdir(path):
            extension = "    " if i == len(entries) - 1 else "│   "
            tree_lines.extend(generate_tree(path, prefix + extension))

    return tree_lines

# Generate tree
tree_lines = [os.path.basename(base_folder)]
tree_lines.extend(generate_tree(base_folder))

# Save tree to txt
with open(tree_output_file, "w", encoding="utf-8") as f:
    f.write("\n".join(tree_lines))

# Save inventory to CSV
df_inventory = pd.DataFrame(inventory)
df_inventory.to_csv(inventory_output_file, index=False)

print("Done!")
print(f"Tree file saved at: {tree_output_file}")
print(f"Inventory CSV saved at: {inventory_output_file}")
```
