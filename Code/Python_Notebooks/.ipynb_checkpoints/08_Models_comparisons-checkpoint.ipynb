```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
lstm_results_df = pd.read_csv("results/lstm_results.csv")
lstm_results_df
```

```python
# ==========================================================
# BEST EPOCH PER PERIOD + TARGET + HORIZON
# ==========================================================

best_lstm = lstm_results_df.loc[
    lstm_results_df.groupby(
        ["Period", "Target", "Horizon"]
    )["RMSE"].idxmin()
].reset_index(drop=True)

best_lstm

# ==========================================================
# EXPORT LSTM RESULTS TO EXCEL
# ==========================================================

import os

os.makedirs(
    "results",
    exist_ok=True
)

best_lstm.to_csv(
    "results/best_lstm_results.csv",
    index=False
)

print("File exported successfully.")
```

```python
lstm_df = pd.read_csv("results/best_lstm_results.csv")
lstm_df.tail()

```

```python
models_df = pd.read_csv("results/validation_results.csv")
models_df
```

```python
arima_df = pd.read_csv("results/arima_results.csv")
cols = ["MAE", "RMSE", "MAPE", "R2"]

arima_df[cols] = arima_df[cols].apply(
    pd.to_numeric,
    errors="coerce"
)

pd.concat(
    [arima_df.head(), arima_df.tail()]
)
```

```python
# ==========================================================
# REMOVE EPOCH COLUMN FROM LSTM RESULTS
# ==========================================================

lstm_df_no_epoch = lstm_df.drop(
    columns=["Epochs"]
)

# ==========================================================
# MERGE ML + LSTM RESULTS
# ==========================================================

final_results_df = pd.concat(
    [models_df, lstm_df_no_epoch, arima_df],
    ignore_index=True
)



# ==========================================================
# SORT RESULTS
# ==========================================================

final_results_df = final_results_df.sort_values(
    ["Period", "Target", "Horizon", "Model"]
).reset_index(drop=True)
final_results_df
```

```python
final_results_df[final_results_df['R2'] < 0]
```

```python
before = len(final_results_df)

final_results_df = final_results_df[
    final_results_df["R2"] > 0
].reset_index(drop=True)

after = len(final_results_df)

print(f"Removed {before - after} rows")
print(f"Remaining {after} rows")
final_results_df = final_results_df.dropna(
    subset=["MAE", "RMSE", "MAPE", "R2"]
).reset_index(drop=True)

final_results_df
```

```python
period_comparison = (
    final_results_df
    .groupby("Period")
    .agg(
        Count =("RMSE", "count"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean")
    )
    .sort_values("RMSE")
    .round(2)
)

period_comparison
```

```python
period_comparison["RMSE"].plot(
    kind="bar"
)

plt.ylabel("Average RMSE")
plt.title("Historical Period Comparison")
plt.tight_layout()
plt.show()
```

```python
final_results_df = final_results_df[
    final_results_df["Period"] == "FULL"
].reset_index(drop=True)

# print(final_results_df.shape)
final_results_df
```

```python
horizon_comparison = (
    final_results_df
    .groupby("Horizon")
    .agg(
        Count =("RMSE", "count"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean")
    )
    .sort_values("Horizon")
    .round(2)
)

horizon_comparison
```

```python
horizon_comparison["RMSE"].plot(
    marker="o"
)

plt.xlabel("Forecast Horizon (Hours)")
plt.ylabel("Average RMSE")
plt.title("Forecast Horizon vs RMSE")
plt.grid(True)
plt.show()
```

```python
# Best Model Performing Per Horizon

best_per_horizon = (
    final_results_df
    .loc[
        final_results_df.groupby("Horizon")["RMSE"].idxmin()
    ]
    .sort_values("Horizon")
    .reset_index(drop=True)
)

best_per_horizon[
    ["Horizon", "Model", "Target", "Period", "RMSE", "MAE", "R2"]
]
```

```python
target_comparison = (
    final_results_df
    .groupby("Target")
    .agg(
        Count =("RMSE", "count"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean")
    )
    .sort_values("RMSE")
    .round(2)
)

target_comparison
```

```python
final_results_df
```

```python
best_configs = (
    final_results_df.loc[
        final_results_df.groupby(
            ["Period","Target","Horizon"]
        )["RMSE"].idxmin()
    ]
    .sort_values(
        ["Period","Target","Horizon"]
    )
    .reset_index(drop=True)
)

best_configs
print("Columns:")
print(best_configs.columns.tolist())

display(
    pd.concat([
        best_configs.head(),
        best_configs.tail()
    ])
)

print(f"\nShape: {best_configs.shape}")
```

```python
best_model_count = (
    best_configs["Model"]
    .value_counts()
    .reset_index()
)

best_model_count.columns = [
    "Model",
    "Wins"
]

best_model_count
```

```python
lstm_vs_ml = (
    best_configs
    .groupby(["Model"])
    .agg(
       
        Count =("RMSE", "count"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean")
    )
    .sort_values("RMSE")
    .round(2)
)

lstm_vs_ml
```

```python
target_best_models = (
    final_results_df.loc[
        final_results_df.groupby(
            ["Target"]
        )["RMSE"].idxmin()
    ]
)

target_best_models
```

```python
model_ranking = (
    final_results_df
    .groupby("Model")
    .agg(
        Count =("RMSE", "count"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean")
    )
    .sort_values("RMSE")
    .round(2)
)

model_ranking
```

```python
import matplotlib.pyplot as plt

model_ranking["RMSE"].plot(
    kind="bar"
)

plt.ylabel("Average RMSE")
plt.title("Model Comparison")
plt.tight_layout()
plt.show()
```

```python
for target in final_results_df["Target"].unique():

    subset = (
        final_results_df[
            final_results_df["Target"] == target
        ]
        .groupby("Model")["RMSE"]
        .mean()
        .sort_values()
    )

    subset.plot(kind="bar")

    plt.title(target)
    plt.ylabel("Average RMSE")
    plt.tight_layout()
    plt.show()
```
