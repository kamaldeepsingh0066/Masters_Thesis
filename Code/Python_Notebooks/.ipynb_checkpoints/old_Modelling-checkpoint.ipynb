```python
import pandas as pd
import numpy as np

# ---------------------------------------------------
# 1. LOAD MODELING DATASET
# ---------------------------------------------------
df = pd.read_csv("Data_Corpus/final/master_energy_dataset_modeling.csv")

df["datetime"] = pd.to_datetime(df["datetime"])

# ---------------------------------------------------
# 2. SORT (VERY IMPORTANT FOR TIME SERIES)
# ---------------------------------------------------
df = df.sort_values("datetime").reset_index(drop=True)

# ---------------------------------------------------
# 3. CHECK MISSING VALUES
# ---------------------------------------------------
print("Missing values BEFORE cleaning:\n")
print(df.isna().sum().sort_values(ascending=False))

# ---------------------------------------------------
# 4. INTERPOLATE ONLY SAFE CONTINUOUS FEATURES
# ---------------------------------------------------
safe_features = [
    "air_temperature",
    "relative_humidity",
    "wind_speed",
    "wind_dir_sin",
    "wind_dir_cos",
    "price_eur_per_mwh"
]

for col in safe_features:
    if col in df.columns:
        df[col] = df[col].interpolate(method="linear", limit_direction="both")

# ---------------------------------------------------
# 5. SOLAR EDGE CASE (small gaps)
# ---------------------------------------------------
solar_cols = [
    "atmospheric_radiation",
    "diffuse_radiation",
    "global_radiation",
    "sunshine_duration",
    "solar_zenith_angle"
]

for col in solar_cols:
    if col in df.columns:
        df[col] = df[col].interpolate(method="linear", limit_direction="both")

# ---------------------------------------------------
# 6. FINAL CHECK
# ---------------------------------------------------
print("\nMissing values AFTER cleaning:\n")
print(df.isna().sum().sort_values(ascending=False))

print("\nDataset shape:", df.shape)

print("\nSample rows:")
print(df.sample(5))



# ---------------------------------------------------
# 7. TIME FEATURE ENGINEERING
# ---------------------------------------------------

# Basic time features
df["year"] = df["datetime"].dt.year
df["month"] = df["datetime"].dt.month
df["day"] = df["datetime"].dt.day
df["hour"] = df["datetime"].dt.hour
df["day_of_week"] = df["datetime"].dt.dayofweek
df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)

# ---------------------------------------------------
# 8. CYCLICAL ENCODING (VERY IMPORTANT)
# ---------------------------------------------------

# Hour (captures daily solar cycle)
df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24)
df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24)

# Month (captures seasonal solar/wind pattern)
df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)

# Day of week (optional but useful)
df["dow_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["dow_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7)

# ---------------------------------------------------
# 9. CHECK
# ---------------------------------------------------
print("\nTime features added successfully")

print("\nSample with time features:")
print(df[[
    "datetime", "hour", "hour_sin", "hour_cos",
    "month", "month_sin", "month_cos",
    "day_of_week", "is_weekend"
]].sample(5))
print(df.shape)


# ---------------------------------------------------
# 10. LAG FEATURE ENGINEERING
# ---------------------------------------------------

target_cols = [
    "generation_photovoltaics_mwh",
    "generation_wind_onshore_mwh",
    "generation_wind_offshore_mwh"
]

lag_hours = [1, 2, 3, 6, 12, 24, 48, 72, 168]

for target in target_cols:
    for lag in lag_hours:
        df[f"{target}_lag_{lag}"] = df[target].shift(lag)

# ---------------------------------------------------
# 11. CHECK NEW MISSING VALUES FROM LAGS
# ---------------------------------------------------
print("\nLag features added successfully")

lag_cols = [col for col in df.columns if "_lag_" in col]

print("\nTotal lag feature columns created:", len(lag_cols))
print("\nMissing values caused by lags:")
print(df[lag_cols].isna().sum().sort_values(ascending=False).head(20))

print("\nSample lag columns:")
print(df[[
    "datetime",
    "generation_photovoltaics_mwh",
    "generation_photovoltaics_mwh_lag_1",
    "generation_photovoltaics_mwh_lag_24",
    "generation_wind_onshore_mwh",
    "generation_wind_onshore_mwh_lag_1",
    "generation_wind_onshore_mwh_lag_24"
]].sample(5))



print(df.shape)
print(df[[col for col in df.columns if "_lag_" in col]].isna().sum().sort_values(ascending=False).head(10))



# ---------------------------------------------------
# 12. ROLLING FEATURE ENGINEERING
# ---------------------------------------------------

rolling_windows = [24, 168]

for target in target_cols:
    for window in rolling_windows:
        df[f"{target}_rolling_mean_{window}"] = df[target].shift(1).rolling(window=window).mean()
        df[f"{target}_rolling_std_{window}"] = df[target].shift(1).rolling(window=window).std()

# ---------------------------------------------------
# 13. CHECK ROLLING FEATURES
# ---------------------------------------------------
print("\nRolling features added successfully")

rolling_cols = [col for col in df.columns if "_rolling_" in col]

print("\nTotal rolling feature columns created:", len(rolling_cols))

print("\nMissing values caused by rolling features:")
print(df[rolling_cols].isna().sum().sort_values(ascending=False).head(20))

print("\nSample rolling features:")
print(df[[
    "datetime",
    "generation_photovoltaics_mwh",
    "generation_photovoltaics_mwh_rolling_mean_24",
    "generation_photovoltaics_mwh_rolling_std_24",
    "generation_wind_onshore_mwh",
    "generation_wind_onshore_mwh_rolling_mean_24"
]].sample(5))

print(df.shape)

print(df[[col for col in df.columns if "_rolling_" in col]].isna().sum().sort_values(ascending=False).head(10))



# ---------------------------------------------------
# 14. DROP STARTUP ROWS CAUSED BY LAGS / ROLLING FEATURES
# ---------------------------------------------------
before_drop_shape = df.shape

df = df.dropna().reset_index(drop=True)

after_drop_shape = df.shape

print("\nDropped startup rows successfully")
print("Before drop:", before_drop_shape)
print("After drop:", after_drop_shape)
print("Rows removed:", before_drop_shape[0] - after_drop_shape[0])

# ---------------------------------------------------
# 15. FINAL CHECK
# ---------------------------------------------------
print("\nFinal missing values:")
print(df.isna().sum().sort_values(ascending=False).head(20))

print("\nFinal dataset shape:", df.shape)

print("\nSample final rows:")
print(df.sample(5))



# ---------------------------------------------------
# 16. SAVE FINAL FEATURE-ENGINEERED DATASET
# ---------------------------------------------------
output_path = "Data_Corpus/final/master_energy_dataset_features.csv"
df.to_csv(output_path, index=False)

print(f"\nFeature-engineered dataset saved successfully to: {output_path}")


print(df.shape)
print(df.isna().sum().sum())





```
