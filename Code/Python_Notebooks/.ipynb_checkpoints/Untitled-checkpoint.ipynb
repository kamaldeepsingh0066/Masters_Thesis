```python
# Temperature Data Historical
import requests
from bs4 import BeautifulSoup
import os

base_url = "https://opendata.dwd.de/climate_environment/CDC/observations_germany/climate/hourly/air_temperature/historical/"

response = requests.get(base_url)
soup = BeautifulSoup(response.text, "html.parser")

links = [a.get("href") for a in soup.find_all("a") if a.get("href").endswith((".zip",".pdf",".txt"))]
os.makedirs("Document_Corpus", exist_ok=True)
os.makedirs("Document_Corpus/dwd_temp", exist_ok=True)
os.makedirs("Document_Corpus/dwd_temp/historical_zip", exist_ok=True)
for link in links:
    file_url = base_url + link
    file_path = os.path.join("Document_Corpus/dwd_temp/historical_zip", link)
    
    print("Downloading:", link)
    r = requests.get(file_url)
    with open(file_path, "wb") as f:
        f.write(r.content)

```
