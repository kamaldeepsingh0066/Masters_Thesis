```python
import pandas as pd
import numpy as np
import os

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping

# ==========================================================
# 1. LOAD DATA
# ==========================================================
df = pd.read_csv("Data_Corpus/final/master_energy_dataset_modeling.csv")
df["datetime"] = pd.to_datetime(df["datetime"])
df = df.sort_values("datetime").reset_index(drop=True)
df = df[df["datetime"] >= "2020-01-01"].reset_index(drop=True)

print("Dataset loaded:", df.shape)

# ==========================================================
# 2. SETTINGS
# ==========================================================
target_map = {
    "solar": "generation_photovoltaics_mwh",
    "wind_onshore": "generation_wind_onshore_mwh",
    "wind_offshore": "generation_wind_offshore_mwh"
}

forecast_horizons = [1, 6, 24, 72, 168]
sequence_length = 24

results = []

# ==========================================================
# 3. FUNCTION TO CREATE SEQUENCES
# ==========================================================
def create_sequences(X, y, seq_len):
    Xs, ys = [], []
    for i in range(seq_len, len(X)):
        Xs.append(X[i-seq_len:i])
        ys.append(y[i])
    return np.array(Xs), np.array(ys)

# ==========================================================
# 4. LOOP THROUGH TARGETS AND HORIZONS
# ==========================================================
for selected_target, target_col in target_map.items():
    for horizon in forecast_horizons:

        print("\n" + "="*80)
        print(f"RUNNING LSTM → TARGET: {selected_target.upper()} | HORIZON: {horizon}h")
        print("="*80)

        # ---------------------------------------------------
        # A. CREATE FUTURE TARGET
        # ---------------------------------------------------
        df_model = df.copy()
        df_model["target"] = df_model[target_col].shift(-horizon)
        df_model = df_model.dropna().reset_index(drop=True)

        # ---------------------------------------------------
        # B. DROP OTHER TARGET GENERATION COLUMNS
        # ---------------------------------------------------
        generation_cols = [
            "generation_photovoltaics_mwh",
            "generation_wind_onshore_mwh",
            "generation_wind_offshore_mwh"
        ]

        for col in generation_cols:
            if col != target_col and col in df_model.columns:
                df_model = df_model.drop(columns=[col])

        print("Modeling dataset shape:", df_model.shape)

        # ---------------------------------------------------
        # C. TRAIN / TEST SPLIT (TIME-BASED)
        # ---------------------------------------------------
        split_index = int(len(df_model) * 0.8)

        train_df = df_model.iloc[:split_index]
        test_df = df_model.iloc[split_index:]

        dt_test = test_df["datetime"].copy()

        X_train = train_df.drop(columns=["datetime", "target"])
        y_train = train_df["target"]

        X_test = test_df.drop(columns=["datetime", "target"])
        y_test = test_df["target"]

        print("Train shape:", X_train.shape, y_train.shape)
        print("Test shape:", X_test.shape, y_test.shape)
        print("Train period:", train_df["datetime"].min(), "→", train_df["datetime"].max())
        print("Test period: ", test_df["datetime"].min(), "→", test_df["datetime"].max())

        # ---------------------------------------------------
        # D. SCALE FEATURES + TARGET
        # ---------------------------------------------------
        scaler_X = MinMaxScaler()
        scaler_y = MinMaxScaler()

        X_train_scaled = scaler_X.fit_transform(X_train)
        X_test_scaled = scaler_X.transform(X_test)

        y_train_scaled = scaler_y.fit_transform(y_train.values.reshape(-1, 1))
        y_test_scaled = scaler_y.transform(y_test.values.reshape(-1, 1))

        # ---------------------------------------------------
        # E. CREATE SEQUENCES
        # ---------------------------------------------------
        X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_scaled, sequence_length)
        X_test_seq, y_test_seq = create_sequences(X_test_scaled, y_test_scaled, sequence_length)

        dt_test_seq = dt_test.iloc[sequence_length:].reset_index(drop=True)

        print("\nSequence shapes:")
        print("X_train_seq:", X_train_seq.shape)
        print("y_train_seq:", y_train_seq.shape)
        print("X_test_seq:", X_test_seq.shape)
        print("y_test_seq:", y_test_seq.shape)

        # ---------------------------------------------------
        # F. BUILD LSTM MODEL
        # ---------------------------------------------------
        model = Sequential([
            Input(shape=(sequence_length, X_train_seq.shape[2])),

            LSTM(64, return_sequences=True),
            Dropout(0.2),

            LSTM(32),
            Dropout(0.2),

            Dense(1)
        ])

        model.compile(optimizer="adam", loss="mse")

        early_stop = EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True
        )

        # ---------------------------------------------------
        # G. TRAIN MODEL
        # ---------------------------------------------------
        history = model.fit(
            X_train_seq,
            y_train_seq,
            epochs=20,
            batch_size=32,
            validation_split=0.1,
            callbacks=[early_stop],
            verbose=1
        )

        # ---------------------------------------------------
        # H. PREDICT
        # ---------------------------------------------------
        y_pred_scaled = model.predict(X_test_seq)

        y_pred = scaler_y.inverse_transform(y_pred_scaled).flatten()
        y_true = scaler_y.inverse_transform(y_test_seq).flatten()

        # ---------------------------------------------------
        # I. METRICS
        # ---------------------------------------------------
        mae = mean_absolute_error(y_true, y_pred)
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))
        r2 = r2_score(y_true, y_pred)

        mape = np.mean(
            np.abs((y_true - y_pred) / np.where(y_true == 0, 1, y_true))
        ) * 100

        # Solar daytime MAPE
        daytime_mape = np.nan
        if selected_target == "solar":
            mask = y_true > 50
            daytime_mape = np.mean(
                np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])
            ) * 100

        # ---------------------------------------------------
        # J. SAVE RESULT
        # ---------------------------------------------------
        results.append({
            "target": selected_target,
            "horizon_hours": horizon,
            "model": "LSTM",
            "mae": mae,
            "rmse": rmse,
            "mape": mape,
            "r2": r2,
            "daytime_mape_solar_only": daytime_mape
        })

        print("\nRESULTS SUMMARY")
        print(f"LSTM → MAE={mae:.2f}, RMSE={rmse:.2f}, MAPE={mape:.2f}, R²={r2:.4f}")

        if selected_target == "solar":
            print(f"Solar Daytime MAPE (y > 50): {daytime_mape:.2f}%")

        # ---------------------------------------------------
        # K. SAVE PREDICTIONS
        # ---------------------------------------------------
        pred_df = pd.DataFrame({
            "datetime": dt_test_seq,
            "actual": y_true,
            "predicted": y_pred
        })

        pred_dir = "Data_Corpus/results/lstm_predictions"
        os.makedirs(pred_dir, exist_ok=True)

        pred_path = os.path.join(
            pred_dir,
            f"lstm_predictions_{selected_target}_{horizon}h.csv"
        )
        pred_df.to_csv(pred_path, index=False)

        print("Predictions saved to:", pred_path)

# ==========================================================
# 5. SAVE FINAL LSTM RESULTS
# ==========================================================
results_df = pd.DataFrame(results)

output_path = "Data_Corpus/results/lstm_all_results.csv"
results_df.to_csv(output_path, index=False)

print("\n" + "="*80)
print("ALL LSTM EXPERIMENTS COMPLETED SUCCESSFULLY")
print("="*80)
print("Saved to:", output_path)

print("\nFINAL LSTM RESULTS:")
print(results_df)

# ==========================================================
# 6. OPTIONAL: BEST LSTM RESULTS
# ==========================================================
print("\nBEST LSTM RESULT PER TARGET:")
best_lstm = results_df.loc[results_df.groupby("target")["rmse"].idxmin()]
print(best_lstm)
```
