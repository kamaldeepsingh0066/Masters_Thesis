```python
import pandas as pd
import numpy as np
import os

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input

# ==========================================================
# 1. LOAD DATA
# ==========================================================
df = pd.read_csv("Data_Corpus/final/master_energy_dataset_modeling.csv")

df["datetime"] = pd.to_datetime(df["datetime"])
df = df.sort_values("datetime").reset_index(drop=True)

df = df[df["global_radiation"] > 0]

print("Full Dataset loaded:", df.shape)
print("Full Date Range:", df["datetime"].min(), "→", df["datetime"].max())

periods = {
    "before_2020": df[df["datetime"] < "2020-01-01"].copy(),
    "from_2021": df[df["datetime"] >= "2021-01-01"].copy(),
    "full_dataset": df.copy()
}

# ==========================================================
# 2. SETTINGS
# ==========================================================
target_map = {
    "solar": "generation_photovoltaics_mwh",
    "wind_onshore": "generation_wind_onshore_mwh",
    "wind_offshore": "generation_wind_offshore_mwh"
}

forecast_horizons = [1,6,24,72,168]
sequence_length = 24

# Epoch experiments
epoch_list = [20,40,60,80,100]

results = []

# ==========================================================
# 3. FUNCTION TO CREATE SEQUENCES
# ==========================================================
def create_sequences(X, y, seq_len):
    Xs, ys = [], []

    for i in range(seq_len, len(X)):
        Xs.append(X[i-seq_len:i])
        ys.append(y[i])

    return np.array(Xs), np.array(ys)

# ==========================================================
# 4. LOOP THROUGH TARGETS, HORIZONS AND EPOCHS
# ==========================================================
for period_name, df_period in periods.items():

    print("\n" + "=" * 100)
    print(f"PERIOD: {period_name}")
    print(
        "Date Range:",
        df_period["datetime"].min(),
        "→",
        df_period["datetime"].max()
    )
    print("=" * 100)

    for selected_target, target_col in target_map.items():

        for horizon in forecast_horizons:

            for epochs_to_run in epoch_list:

                print("\n" + "=" * 100)
                print(
                    f"RUNNING LSTM | TARGET={selected_target.upper()} "
                    f"| HORIZON={horizon}h "
                    f"| EPOCHS={epochs_to_run}"
                )
                print("=" * 100)
    
                # -------------------------------------------------
                # A. CREATE FUTURE TARGET
                # -------------------------------------------------
                df_model = df_period.copy()
    
                df_model["target"] = df_model[target_col].shift(-horizon)
                # REMOVE NANs CREATED BY SHIFT
                df_model = df_model.dropna().reset_index(drop=True)
    
                if len(df_model) < 1000:
                    
                    print(
                        f"Skipping {period_name} | "
                        f"{selected_target} | "
                        f"{horizon}h : insufficient data"
                    )
                    continue
    
                # -------------------------------------------------
                # B. DROP OTHER GENERATION TARGETS
                # -------------------------------------------------
                generation_cols = [
                    "generation_photovoltaics_mwh",
                    "generation_wind_onshore_mwh",
                    "generation_wind_offshore_mwh"
                ]
    
                for col in generation_cols:
                    if col != target_col and col in df_model.columns:
                        df_model = df_model.drop(columns=[col])
    
                print("Modeling dataset shape:", df_model.shape)
    
                # -------------------------------------------------
                # C. TRAIN TEST SPLIT
                # -------------------------------------------------
                split_index = int(len(df_model) * 0.80)
    
                train_df = df_model.iloc[:split_index]
                test_df = df_model.iloc[split_index:]
    
                dt_test = test_df["datetime"].copy()
    
                X_train = train_df.drop(columns=["datetime", "target"])
                y_train = train_df["target"]
    
                X_test = test_df.drop(columns=["datetime", "target"])
                y_test = test_df["target"]
    
                print("Train shape:", X_train.shape, y_train.shape)
                print("Test shape:", X_test.shape, y_test.shape)
    
                print(
                    "Train period:",
                    train_df["datetime"].min(),
                    "→",
                    train_df["datetime"].max()
                )
    
                print(
                    "Test period:",
                    test_df["datetime"].min(),
                    "→",
                    test_df["datetime"].max()
                )
    
                # -------------------------------------------------
                # D. SCALING
                # -------------------------------------------------
                scaler_X = MinMaxScaler()
                scaler_y = MinMaxScaler()
    
                X_train_scaled = scaler_X.fit_transform(X_train)
                X_test_scaled = scaler_X.transform(X_test)
    
                y_train_scaled = scaler_y.fit_transform(
                    y_train.values.reshape(-1, 1)
                )
    
                y_test_scaled = scaler_y.transform(
                    y_test.values.reshape(-1, 1)
                )
    
                # -------------------------------------------------
                # E. CREATE SEQUENCES
                # -------------------------------------------------
                X_train_seq, y_train_seq = create_sequences(
                    X_train_scaled,
                    y_train_scaled,
                    sequence_length
                )
    
                X_test_seq, y_test_seq = create_sequences(
                    X_test_scaled,
                    y_test_scaled,
                    sequence_length
                )
    
                dt_test_seq = (
                    dt_test.iloc[sequence_length:]
                    .reset_index(drop=True)
                )
    
                print("\nSequence Shapes")
                print("X_train_seq:", X_train_seq.shape)
                print("y_train_seq:", y_train_seq.shape)
                print("X_test_seq:", X_test_seq.shape)
                print("y_test_seq:", y_test_seq.shape)
    
                # -------------------------------------------------
                # F. BUILD MODEL
                # -------------------------------------------------
                model = Sequential([
                    Input(shape=(sequence_length, X_train_seq.shape[2])),
    
                    LSTM(
                        64,
                        return_sequences=True
                    ),
                    Dropout(0.2),
    
                    LSTM(32),
                    Dropout(0.2),
    
                    Dense(1)
                ])
    
                model.compile(
                    optimizer="adam",
                    loss="mse"
                )
    
                # -------------------------------------------------
                # G. TRAIN MODEL
                # -------------------------------------------------
                history = model.fit(
                    X_train_seq,
                    y_train_seq,
                    epochs=epochs_to_run,
                    batch_size=32,
                    validation_split=0.10,
                    verbose=1
                )
    
                # -------------------------------------------------
                # H. PREDICT
                # -------------------------------------------------
                y_pred_scaled = model.predict(X_test_seq)
    
                y_pred = scaler_y.inverse_transform(
                    y_pred_scaled
                ).flatten()
    
                y_true = scaler_y.inverse_transform(
                    y_test_seq
                ).flatten()
    
                # -------------------------------------------------
                # I. METRICS
                # -------------------------------------------------
                mae = mean_absolute_error(
                    y_true,
                    y_pred
                )
    
                rmse = np.sqrt(
                    mean_squared_error(
                        y_true,
                        y_pred
                    )
                )
    
                r2 = r2_score(
                    y_true,
                    y_pred
                )
    
                mape = np.mean(
                    np.abs(
                        (y_true - y_pred)
                        / np.where(y_true == 0, 1, y_true)
                    )
                ) * 100
    
                daytime_mape = np.nan
    
                if selected_target == "solar":
    
                    mask = y_true > 50
    
                    if mask.sum() > 0:
                        daytime_mape = np.mean(
                            np.abs(
                                (
                                    y_true[mask]
                                    - y_pred[mask]
                                )
                                / y_true[mask]
                            )
                        ) * 100
    
                # -------------------------------------------------
                # J. SAVE RESULTS
                # -------------------------------------------------
                results.append({
                    
                    "period": period_name,
                    "train_start": train_df["datetime"].min(),
                    "train_end": train_df["datetime"].max(),
                    "test_start": test_df["datetime"].min(),
                    "test_end": test_df["datetime"].max(),

                    "target": selected_target,
                    "horizon_hours": horizon,
                    "epochs": epochs_to_run,
                    "model": "LSTM",
                    "mae": mae,
                    "rmse": rmse,
                    "mape": mape,
                    "r2": r2,
                    "daytime_mape_solar_only": daytime_mape
                })
    
                print("\nRESULTS SUMMARY")
                print(
                    f"MAE={mae:.4f} | "
                    f"RMSE={rmse:.4f} | "
                    f"MAPE={mape:.4f}% | "
                    f"R²={r2:.4f}"
                )
    
                if selected_target == "solar":
                    print(
                        f"Solar Daytime MAPE: "
                        f"{daytime_mape:.4f}%"
                    )
    
                # -------------------------------------------------
                # K. SAVE PREDICTIONS
                # -------------------------------------------------
                pred_df = pd.DataFrame({
                    "datetime": dt_test_seq,
                    "actual": y_true,
                    "predicted": y_pred
                })
    
                pred_dir = (
                    "Data_Corpus/results/"
                    "lstm_predictions_globalradiation"
                )
    
                os.makedirs(
                    pred_dir,
                    exist_ok=True
                )
    
                pred_path = os.path.join(
                    pred_dir,
                    f"{period_name}_"
                    f"{selected_target}_"
                    f"{horizon}h_"
                    f"{epochs_to_run}epochs.csv"
                )
    
                pred_df.to_csv(
                    pred_path,
                    index=False
                )
    
                print(
                    "Predictions saved:",
                    pred_path
                )

# ==========================================================
# 5. SAVE ALL RESULTS
# ==========================================================
results_df = pd.DataFrame(results)

output_path = (
    "Data_Corpus/results/"
    "lstm_all_results_epochs_comparison_globalradiation.csv"
)

results_df.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 100)
print("ALL EXPERIMENTS COMPLETED")
print("=" * 100)

print("Saved:", output_path)

# ==========================================================
# 6. DISPLAY ALL RESULTS
# ==========================================================
print("\n" + "=" * 100)
print("ALL EXPERIMENT RESULTS")
print("=" * 100)

print(
    results_df[
        [
            "period",
            "target",
            "horizon_hours",
            "epochs",
            "rmse",
            "mae",
            "mape",
            "daytime_mape_solar_only",
            "r2"
        ]
    ].sort_values(
        by=["period", "target", "rmse"]
    )
)

# ==========================================================
# 7. BEST MODEL PER TARGET
# ==========================================================
print("\n\n\nBEST LSTM RESULT PER PERIOD AND TARGET")

best_lstm = results_df.loc[
    results_df.groupby(
        ["period", "target"]
    )["rmse"].idxmin()
]

print(
    best_lstm[
        [
            "period",
            "target",
            "horizon_hours",
            "epochs",
            "rmse",
            "mae",
            "mape",
            "daytime_mape_solar_only",
            "r2"
        ]
    ]
)
```
