# 08 - Model Comparison (corrected)

Changes from the original notebook:
- Removed the `R2 > 0` filter and the `Period == "FULL"` filter. Both silently discarded model results rather than data errors, and did so unevenly across models (models that fail often had more of their results removed than models that rarely fail). See the aggregate tables below for why this matters.
- Added a `Success` column (`R2 > 0`) so failure rate is visible and reportable instead of hidden.
- Per-config selection tables (`best_configs`, `best_model_count`, `lstm_vs_ml`, `target_best_models`, `best_per_horizon`) are unchanged in logic - they were never distorted by the filters, since a failing model can't win a slot it's already losing. They now simply cover the full 45-config grid (3 periods x 3 targets x 5 horizons) instead of the 15-config FULL-only subset.
- Aggregate characterization tables (`model_ranking`, `horizon_comparison`, `target_comparison`, `period_comparison`) now report `Success_Rate` alongside the mean, since these tables make a general claim about a model's/horizon's/period's typical performance, and that claim isn't supportable if failures are excluded.

```
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```
lstm_results_df = pd.read_csv("results/lstm_results.csv")
lstm_results_df
```

```
# ==========================================================
# BEST EPOCH PER PERIOD + TARGET + HORIZON
# ==========================================================
best_lstm = lstm_results_df.loc[
    lstm_results_df.groupby(["Period", "Target", "Horizon"])["RMSE"].idxmin()
].reset_index(drop=True)
best_lstm

# ==========================================================
# EXPORT LSTM RESULTS TO EXCEL
# ==========================================================
os.makedirs("results", exist_ok=True)
best_lstm.to_csv("results/best_lstm_results.csv", index=False)
print("File exported successfully.")
```

```
lstm_df = pd.read_csv("results/best_lstm_results.csv")
lstm_df.tail()
```

```
models_df = pd.read_csv("results/validation_results.csv")
models_df
```

```
arima_df = pd.read_csv("results/arima_results.csv")
cols = ["MAE", "RMSE", "MAPE", "R2"]
arima_df[cols] = arima_df[cols].apply(pd.to_numeric, errors="coerce")
pd.concat([arima_df.head(), arima_df.tail()])
```

## Merge (unchanged)

```
# ==========================================================
# REMOVE EPOCH COLUMN FROM LSTM RESULTS
# ==========================================================
lstm_df_no_epoch = lstm_df.drop(columns=["Epochs"])

# ==========================================================
# MERGE ML + LSTM + ARIMA RESULTS
# ==========================================================
final_results_df = pd.concat(
    [models_df, lstm_df_no_epoch, arima_df], ignore_index=True
)

# ==========================================================
# SORT RESULTS
# ==========================================================
final_results_df = final_results_df.sort_values(
    ["Period", "Target", "Horizon", "Model"]
).reset_index(drop=True)

print(f"Total experiments loaded: {len(final_results_df)}")
final_results_df
```

```
# Inspect negative-R2 rows directly - kept for visibility, nothing removed here
final_results_df[final_results_df["R2"] < 0]
```

## Success flag (new)

This replaces the old `R2 > 0` filter. Instead of deleting rows, we label them, so every aggregate table below can report a success rate instead of silently excluding failures.

```
final_results_df["Success"] = final_results_df["R2"] > 0
print(f"Overall success rate: {final_results_df['Success'].mean()*100:.1f}%  "
      f"({final_results_df['Success'].sum()} of {len(final_results_df)} configs)")
```

## Period comparison (corrected: no filtering, success rate added)

**Note on interpretation:** `FULL` validates on 2023 while `POST_2020`/`POST_2021` validate on 2024 (different test years, not just different training-window lengths). This table is descriptive only and should not be presented as isolating a pandemic or training-window effect without noting that confound.

```
period_comparison = (
    final_results_df
    .groupby("Period")
    .agg(
        Count=("RMSE", "count"),
        Success_Rate=("Success", "mean"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean"),
    )
    .sort_values("RMSE")
    .round(2)
)
period_comparison["Success_Rate"] = (period_comparison["Success_Rate"] * 100).round(1)
period_comparison
```

```
period_comparison["RMSE"].plot(kind="bar")
plt.ylabel("Average RMSE")
plt.title("Historical Period Comparison (all configs, success rate shown above)")
plt.tight_layout()
plt.show()
```

## Horizon comparison (corrected: no filtering, success rate added)

```
horizon_comparison = (
    final_results_df
    .groupby("Horizon")
    .agg(
        Count=("RMSE", "count"),
        Success_Rate=("Success", "mean"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean"),
    )
    .sort_values("Horizon")
    .round(2)
)
horizon_comparison["Success_Rate"] = (horizon_comparison["Success_Rate"] * 100).round(1)
horizon_comparison
```

```
horizon_comparison["RMSE"].plot(marker="o")
plt.xlabel("Forecast Horizon (Hours)")
plt.ylabel("Average RMSE")
plt.title("Forecast Horizon vs RMSE (all configs)")
plt.grid(True)
plt.show()
```

## Best model per horizon (unchanged logic - per-config selection, not affected by the filters)

```
# Best Model Performing Per Horizon
best_per_horizon = (
    final_results_df
    .loc[final_results_df.groupby("Horizon")["RMSE"].idxmin()]
    .sort_values("Horizon")
    .reset_index(drop=True)
)
best_per_horizon[["Horizon", "Model", "Target", "Period", "RMSE", "MAE", "R2"]]
```

## Target comparison (corrected: no filtering, success rate added)

```
target_comparison = (
    final_results_df
    .groupby("Target")
    .agg(
        Count=("RMSE", "count"),
        Success_Rate=("Success", "mean"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean"),
    )
    .sort_values("RMSE")
    .round(2)
)
target_comparison["Success_Rate"] = (target_comparison["Success_Rate"] * 100).round(1)
target_comparison
```

```
final_results_df
```

## Best config per (Period, Target, Horizon) - unchanged logic

Per-config winner selection. A failing model can never win a slot another model succeeds in, so this table was never distorted by the old filters - it now simply covers all 45 configs instead of only the 15 in the `FULL`-only subset.

**Check this yourself:** confirm `best_configs.shape[0]` equals the number of (Period, Target, Horizon) combinations you actually ran (3 x 3 x 5 = 45 if all three periods use the same target/horizon grid). If it's fewer, at least one combination had every model fail simultaneously (all R2 <= 0) - `idxmin` will still pick something in that case since it doesn't drop rows, so a shortfall here would only happen if a combination is missing from the data entirely, not because of the Success filter.

```
best_configs = (
    final_results_df
    .loc[final_results_df.groupby(["Period", "Target", "Horizon"])["RMSE"].idxmin()]
    .sort_values(["Period", "Target", "Horizon"])
    .reset_index(drop=True)
)
print("Columns:")
print(best_configs.columns.tolist())
display(pd.concat([best_configs.head(), best_configs.tail()]))
print(f"\nShape: {best_configs.shape}")
```

```
best_model_count = (
    best_configs["Model"].value_counts().reset_index()
)
best_model_count.columns = ["Model", "Wins"]
print(f"Total wins: {best_model_count['Wins'].sum()} "
      f"(should equal number of Period x Target x Horizon combinations)")
best_model_count
```

```
lstm_vs_ml = (
    best_configs
    .groupby(["Model"])
    .agg(
        Count=("RMSE", "count"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean"),
    )
    .sort_values("RMSE")
    .round(2)
)
lstm_vs_ml
```

```
target_best_models = (
    final_results_df
    .loc[final_results_df.groupby(["Target"])["RMSE"].idxmin()]
)
target_best_models
```

## Overall model ranking (corrected: no filtering, success rate added)

This is the headline table - report `Success_Rate` next to `RMSE`/`R2` always. A low success rate with a good conditional average (e.g. ARIMA-family) means the model rarely works, not that it's reliable.

```
model_ranking = (
    final_results_df
    .groupby("Model")
    .agg(
        Count=("RMSE", "count"),
        Success_Rate=("Success", "mean"),
        RMSE=("RMSE", "mean"),
        MAE=("MAE", "mean"),
        R2=("R2", "mean"),
    )
    .sort_values("RMSE")
    .round(2)
)
model_ranking["Success_Rate"] = (model_ranking["Success_Rate"] * 100).round(1)
model_ranking
```

### Optional: conditional performance (only among successful configs)

Always report this next to `Success_Rate` above, never in place of it.

```
conditional_ranking = (
    final_results_df[final_results_df["Success"]]
    .groupby("Model")
    .agg(
        Successful_Count=("RMSE", "count"),
        RMSE_when_successful=("RMSE", "mean"),
        R2_when_successful=("R2", "mean"),
    )
    .sort_values("RMSE_when_successful")
    .round(2)
)
model_ranking.join(conditional_ranking, how="left")
```

```
model_ranking["RMSE"].plot(kind="bar")
plt.ylabel("Average RMSE")
plt.title("Model Comparison (all configs)")
plt.tight_layout()
plt.show()
```

```
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
model_ranking["RMSE"].plot(kind="bar", ax=axes[0], color="steelblue")
axes[0].set_title("Average RMSE (all configs)")
axes[0].tick_params(axis="x", rotation=45)
model_ranking["Success_Rate"].plot(kind="bar", ax=axes[1], color="indianred")
axes[1].set_title("Success Rate (% configs with R2 > 0)")
axes[1].set_ylabel("Success Rate (%)")
axes[1].tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show()
```

## Per-target model comparison (unchanged)

```
for target in final_results_df["Target"].unique():
    subset = (
        final_results_df[final_results_df["Target"] == target]
        .groupby("Model")["RMSE"]
        .mean()
        .sort_values()
    )
    subset.plot(kind="bar")
    plt.title(target)
    plt.ylabel("Average RMSE")
    plt.tight_layout()
    plt.show()
```
