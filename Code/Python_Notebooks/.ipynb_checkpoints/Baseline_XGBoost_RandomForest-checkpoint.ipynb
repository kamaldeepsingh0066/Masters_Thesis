```python
import pandas as pd
import numpy as np
import os

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
import os

# ==========================================================
# 1. LOAD DATA
# ==========================================================
df = pd.read_csv("Data_Corpus/final/master_energy_dataset_modeling.csv")

df["datetime"] = pd.to_datetime(df["datetime"])
df = df.sort_values("datetime").reset_index(drop=True)

print("Full Dataset loaded:", df.shape)
print("Full Date Range:", df["datetime"].min(), "→", df["datetime"].max())

# ---------------------------------------------------
#  CREATE FORECAST HORIZON TARGETS
# ---------------------------------------------------
forecast_targets = [
    "generation_photovoltaics_mwh",
    "generation_wind_onshore_mwh",
    "generation_wind_offshore_mwh"
]

forecast_horizons = [1, 6, 24, 72, 168]

for target in forecast_targets:
    for horizon in forecast_horizons:
        df[f"{target}_target_t_plus_{horizon}"] = df[target].shift(-horizon)

# ==========================================================
# CREATE PERIODS AFTER TARGETS EXIST
# ==========================================================
periods = {
    "before_2020": df[df["datetime"] < "2020-01-01"].copy(),
    "from_2021": df[df["datetime"] >= "2021-01-01"].copy(),
    "full_dataset": df.copy()
}

for period_name, df_period in periods.items():
    print("\n" + "="*80)
    print(period_name)
    print("Rows:", len(df_period))
    print(
        "Range:",
        df_period["datetime"].min(),
        "→",
        df_period["datetime"].max()
    )


# ---------------------------------------------------
# 18. CHECK NEW TARGET COLUMNS
# ---------------------------------------------------
target_cols_created = [col for col in df.columns if "_target_t_plus_" in col]

print("\nForecast target columns created successfully")
print("Total new target columns:", len(target_cols_created))

print("\nMissing values caused by future target shifting:")
print(df[target_cols_created].isna().sum().sort_values(ascending=False).head(20))

print("\nSample future targets:")
print(df[[
    "datetime",
    "generation_photovoltaics_mwh",
    "generation_photovoltaics_mwh_target_t_plus_1",
    "generation_photovoltaics_mwh_target_t_plus_24",
    "generation_wind_onshore_mwh",
    "generation_wind_onshore_mwh_target_t_plus_1",
    "generation_wind_onshore_mwh_target_t_plus_24"
]].sample(5))
print(df.shape)
print(df[[col for col in df.columns if "_target_t_plus_" in col]].isna().sum().sort_values(ascending=False).head(10))

# ===================================================
# 19. AUTOMATED EXPERIMENT LOOP
# All targets × all horizons × Baseline / RF / XGB
# ===================================================

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
import os

# ---------------------------------------------------
# SETTINGS
# ---------------------------------------------------
target_map = {
    "solar": "generation_photovoltaics_mwh",
    "wind_onshore": "generation_wind_onshore_mwh",
    "wind_offshore": "generation_wind_offshore_mwh"
}

forecast_horizons = [1, 6, 24, 72, 168]

results = []
feature_importance_results = []

# ---------------------------------------------------
# LOOP THROUGH PERIODS, TARGETS AND HORIZONS
# ---------------------------------------------------
for period_name, df_period in periods.items():

    print("\n" + "="*100)
    print(f"PERIOD: {period_name}")
    print(
        "Date Range:",
        df_period["datetime"].min(),
        "→",
        df_period["datetime"].max()
    )
    print("="*100)

    for selected_target, base_target_col in target_map.items():

        for horizon in forecast_horizons:
            print("\n" + "="*80)
            print(f"RUNNING EXPERIMENT → TARGET: {selected_target.upper()} | HORIZON: {horizon}h")
            print("="*80)
    
            # ---------------------------------------------------
            # A. BUILD TARGET COLUMN
            # ---------------------------------------------------
            target_col = f"{base_target_col}_target_t_plus_{horizon}"
    
            # ---------------------------------------------------
            # B. CREATE TARGET-SPECIFIC DATASET
            # ---------------------------------------------------
            df_model = df_period.dropna( subset=[target_col] ).copy()
    
            # Drop all future target columns except selected one
            future_target_cols = [col for col in df_model.columns if "_target_t_plus_" in col]
            cols_to_drop = [col for col in future_target_cols if col != target_col]
            df_model = df_model.drop(columns=cols_to_drop)
    
            # Rename selected future target to generic target
            df_model = df_model.rename(columns={target_col: "target"})
    
            # Drop other current targets (optional but cleaner)
            if selected_target == "wind_onshore":
                df_model = df_model.drop(columns=[
                    "generation_photovoltaics_mwh",
                    "generation_wind_offshore_mwh"
                ], errors="ignore")
    
            elif selected_target == "wind_offshore":
                df_model = df_model.drop(columns=[
                    "generation_photovoltaics_mwh",
                    "generation_wind_onshore_mwh"
                ], errors="ignore")
    
            elif selected_target == "solar":
                df_model = df_model.drop(columns=[
                    "generation_wind_onshore_mwh",
                    "generation_wind_offshore_mwh"
                ], errors="ignore")
    
            print(f"Dataset shape: {df_model.shape}")
    
            # ---------------------------------------------------
            # C. TRAIN / TEST SPLIT (TIME-BASED)
            # ---------------------------------------------------
            datetime_series = df_model["datetime"].copy()
    
            X = df_model.drop(columns=["datetime", "target"])
            y = df_model["target"]
    
            split_index = int(len(df_model) * 0.8)
    
            X_train = X.iloc[:split_index]
            X_test = X.iloc[split_index:]
    
            y_train = y.iloc[:split_index]
            y_test = y.iloc[split_index:]
    
            dt_train = datetime_series.iloc[:split_index]
            dt_test = datetime_series.iloc[split_index:]
    
            print("Train period:", dt_train.min(), "→", dt_train.max())
            print("Test period: ", dt_test.min(), "→", dt_test.max())
    
            # ---------------------------------------------------
            # D. BASELINE MODEL (PERSISTENCE)
            # ---------------------------------------------------
            baseline_feature_map = {
                "solar": "generation_photovoltaics_mwh",
                "wind_onshore": "generation_wind_onshore_mwh",
                "wind_offshore": "generation_wind_offshore_mwh"
            }
    
            baseline_pred = X_test[baseline_feature_map[selected_target]]
    
            baseline_mae = mean_absolute_error(y_test, baseline_pred)
            baseline_rmse = np.sqrt(mean_squared_error(y_test, baseline_pred))
            baseline_r2 = r2_score(y_test, baseline_pred)
            baseline_mape = np.mean(
                np.abs((y_test - baseline_pred) / np.where(y_test == 0, 1, y_test))
            ) * 100
    
            # Solar daytime MAPE only
            baseline_day_mape = np.nan
            if selected_target == "solar":
                mask = y_test > 50
                baseline_day_mape = np.mean(
                    np.abs((y_test[mask] - baseline_pred[mask]) / y_test[mask])
                ) * 100
    
            results.append({
                "period": period_name,
                "target": selected_target,
                "horizon_hours": horizon,
                "model": "Baseline",
                "mae": baseline_mae,
                "rmse": baseline_rmse,
                "mape": baseline_mape,
                "r2": baseline_r2,
                "daytime_mape_solar_only": baseline_day_mape
            })
    
            # ---------------------------------------------------
            # E. RANDOM FOREST
            # ---------------------------------------------------
            rf_model = RandomForestRegressor(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            )
    
            rf_model.fit(X_train, y_train)
            rf_pred = rf_model.predict(X_test)
    
            rf_mae = mean_absolute_error(y_test, rf_pred)
            rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
            rf_r2 = r2_score(y_test, rf_pred)
            rf_mape = np.mean(
                np.abs((y_test - rf_pred) / np.where(y_test == 0, 1, y_test))
            ) * 100
    
            rf_day_mape = np.nan
            if selected_target == "solar":
                mask = y_test > 50
                rf_day_mape = np.mean(
                    np.abs((y_test[mask] - rf_pred[mask]) / y_test[mask])
                ) * 100
    
            results.append({
                "period": period_name,
                "target": selected_target,
                "horizon_hours": horizon,
                "model": "Random Forest",
                "mae": rf_mae,
                "rmse": rf_rmse,
                "mape": rf_mape,
                "r2": rf_r2,
                "daytime_mape_solar_only": rf_day_mape
            })
    
            # ---------------------------------------------------
            # F. XGBOOST
            # ---------------------------------------------------
            xgb_model = XGBRegressor(
                n_estimators=200,
                learning_rate=0.05,
                max_depth=6,
                subsample=0.8,
                colsample_bytree=0.8,
                random_state=42
            )
    
            xgb_model.fit(X_train, y_train)
            xgb_pred = xgb_model.predict(X_test)
    
            xgb_mae = mean_absolute_error(y_test, xgb_pred)
            xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_pred))
            xgb_r2 = r2_score(y_test, xgb_pred)
            xgb_mape = np.mean(
                np.abs((y_test - xgb_pred) / np.where(y_test == 0, 1, y_test))
            ) * 100
    
            xgb_day_mape = np.nan
            if selected_target == "solar":
                mask = y_test > 50
                xgb_day_mape = np.mean(
                    np.abs((y_test[mask] - xgb_pred[mask]) / y_test[mask])
                ) * 100
    
            results.append({
                "period": period_name,
                "target": selected_target,
                "horizon_hours": horizon,
                "model": "XGBoost",
                "mae": xgb_mae,
                "rmse": xgb_rmse,
                "mape": xgb_mape,
                "r2": xgb_r2,
                "daytime_mape_solar_only": xgb_day_mape
            })
    
            # ---------------------------------------------------
            # G. SAVE FEATURE IMPORTANCE (XGBOOST ONLY)
            # ---------------------------------------------------
            xgb_importance = pd.DataFrame({
                "period": period_name,
                "target": selected_target,
                "horizon_hours": horizon,
                "feature": X_train.columns,
                "importance": xgb_model.feature_importances_
            }).sort_values("importance", ascending=False)
    
            feature_importance_results.append(xgb_importance)
    
            # ---------------------------------------------------
            # H. PRINT SUMMARY
            # ---------------------------------------------------
            print("\nRESULTS SUMMARY")
            print(f"Baseline      → MAE={baseline_mae:.2f}, RMSE={baseline_rmse:.2f}, MAPE={baseline_mape:.2f}, R²={baseline_r2:.4f}")
            print(f"Random Forest → MAE={rf_mae:.2f}, RMSE={rf_rmse:.2f}, MAPE={rf_mape:.2f}, R²={rf_r2:.4f}")
            print(f"XGBoost       → MAE={xgb_mae:.2f}, RMSE={xgb_rmse:.2f}, MAPE={xgb_mape:.2f}, R²={xgb_r2:.4f}")
    
            if selected_target == "solar":
                print("\nSolar Daytime MAPE (y > 50)")
                print(f"Baseline      → {baseline_day_mape:.2f}%")
                print(f"Random Forest → {rf_day_mape:.2f}%")
                print(f"XGBoost       → {xgb_day_mape:.2f}%")

# ===================================================
# 20. SAVE FINAL RESULTS
# ===================================================

results_df = pd.DataFrame(results)
feature_importance_df = pd.concat(feature_importance_results, ignore_index=True)

output_dir = "Data_Corpus/results"
os.makedirs(output_dir, exist_ok=True)

results_path = os.path.join(output_dir, "forecast_experiment_results.csv")
feature_path = os.path.join(output_dir, "xgboost_feature_importance_all.csv")

results_df.to_csv(results_path, index=False)
feature_importance_df.to_csv(feature_path, index=False)

print("\n" + "="*80)
print("ALL EXPERIMENTS COMPLETED SUCCESSFULLY")
print("="*80)
print("Results saved to:", results_path)
print("Feature importance saved to:", feature_path)

# ===================================================
# 21. DISPLAY FINAL TABLES
# ===================================================

print("\nFINAL RESULTS TABLE:")
print(results_df.sort_values(["period" , "target", "horizon_hours", "model"]))

print("\nBEST MODEL PER TARGET / HORIZON (Lowest RMSE):")
best_models = results_df.loc[results_df.groupby(["period" , "target", "horizon_hours"])["rmse"].idxmin()]
print(best_models.sort_values(["target", "horizon_hours"]))

# Optional: prettier pivot view
pivot_rmse = results_df.pivot_table(
    index=["target", "horizon_hours"],
    columns="model",
    values="rmse"
)

print("\nRMSE COMPARISON TABLE:")
print(pivot_rmse)
```
