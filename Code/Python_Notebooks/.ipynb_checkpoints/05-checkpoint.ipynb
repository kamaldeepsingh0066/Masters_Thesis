```python
# ==========================================================
# 1. IMPORT LIBRARIES
# ==========================================================

import os
import numpy as np
import pandas as pd

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
import time

start_time = time.time()

# ==========================================================
# 2. LOAD MODELLING DATASETS
# ==========================================================

solar_model_df = pd.read_csv("Data_Corpus/final/modelling/solar_model_dataset.csv")
wind_onshore_model_df = pd.read_csv("Data_Corpus/final/modelling/wind_onshore_model_dataset.csv")
wind_offshore_model_df = pd.read_csv("Data_Corpus/final/modelling/wind_offshore_model_dataset.csv")

solar_model_df["datetime"] = pd.to_datetime(solar_model_df["datetime"])
solar_model_df = solar_model_df.sort_values("datetime").reset_index(drop=True)

wind_onshore_model_df["datetime"] = pd.to_datetime(wind_onshore_model_df["datetime"])
wind_onshore_model_df = wind_onshore_model_df.sort_values("datetime").reset_index(drop=True)

wind_offshore_model_df["datetime"] = pd.to_datetime(wind_offshore_model_df["datetime"])
wind_offshore_model_df = wind_offshore_model_df.sort_values("datetime").reset_index(drop=True)

DATASETS = {
    "Solar": solar_model_df,
    "Wind_Onshore": wind_onshore_model_df,
    "Wind_Offshore": wind_offshore_model_df
}

# ==========================================================
# 3. EXPERIMENT CONFIGURATION
# ==========================================================

PERIODS = {

    "FULL": {
        "start": "2015-01-08 00:00:00",
        "train_end": "2022-12-31 23:00:00",
        "valid_end": "2023-12-31 23:00:00"
    },

    "POST_2020": {
        "start": "2020-01-01 00:00:00",
        "train_end": "2023-12-31 23:00:00",
        "valid_end": "2024-12-31 23:00:00"
    },

    "POST_2021": {
        "start": "2021-01-01 00:00:00",
        "train_end": "2023-12-31 23:00:00",
        "valid_end": "2024-12-31 23:00:00"
    }

}

###################################
TARGETS = {

    "Solar":
    "generation_photovoltaics_mwh",

    "Wind_Onshore":
    "generation_wind_onshore_mwh",

    "Wind_Offshore":
    "generation_wind_offshore_mwh"

}


########################################
HORIZONS = [
    1,
    6,
    24,
    72,
    168
]

##############################################
MODEL_NAMES = [
    "Persistence",
    "RandomForest",
    "XGBoost",
    "LightGBM",
    # "LSTM"
]


# ==========================================================
# 4. FEATURE DEFINITIONS
# ==========================================================

SOLAR_FEATURES = [

    "global_radiation",
    "sunshine_duration",
    "solar_zenith_angle",
    "air_temperature",
    "relative_humidity",

    "capacity_photovoltaics_mw",

    "grid_load_mwh",
    "price_eur_per_mwh",

    "hour",
    "month",
    "dayofweek",

    "generation_photovoltaics_mwh_lag_1",
    "generation_photovoltaics_mwh_lag_6",
    "generation_photovoltaics_mwh_lag_24",
    "generation_photovoltaics_mwh_lag_72",
    "generation_photovoltaics_mwh_lag_168",

    "generation_photovoltaics_mwh_roll_mean_24",
    "generation_photovoltaics_mwh_roll_mean_168",
    "generation_photovoltaics_mwh_roll_std_24"
]
####################################
WIND_ON_FEATURES = [

    "wind_speed",
    "wind_dir_sin",
    "wind_dir_cos",

    "air_temperature",
    "relative_humidity",

    "capacity_wind_onshore_mw",

    "grid_load_mwh",
    "price_eur_per_mwh",

    "hour",
    "month",
    "dayofweek",

    "generation_wind_onshore_mwh_lag_1",
    "generation_wind_onshore_mwh_lag_6",
    "generation_wind_onshore_mwh_lag_24",
    "generation_wind_onshore_mwh_lag_72",
    "generation_wind_onshore_mwh_lag_168",

    "generation_wind_onshore_mwh_roll_mean_24",
    "generation_wind_onshore_mwh_roll_mean_168",
    "generation_wind_onshore_mwh_roll_std_24"
]
##################################

WIND_OFF_FEATURES = [

    "wind_speed",
    "wind_dir_sin",
    "wind_dir_cos",

    "air_temperature",
    "relative_humidity",

    "capacity_wind_offshore_mw",

    "grid_load_mwh",
    "price_eur_per_mwh",

    "hour",
    "month",
    "dayofweek",

    "generation_wind_offshore_mwh_lag_1",
    "generation_wind_offshore_mwh_lag_6",
    "generation_wind_offshore_mwh_lag_24",
    "generation_wind_offshore_mwh_lag_72",
    "generation_wind_offshore_mwh_lag_168",

    "generation_wind_offshore_mwh_roll_mean_24",
    "generation_wind_offshore_mwh_roll_mean_168",
    "generation_wind_offshore_mwh_roll_std_24"
]

##############################################
FEATURES = {
    "Solar": SOLAR_FEATURES,
    "Wind_Onshore": WIND_ON_FEATURES,
    "Wind_Offshore": WIND_OFF_FEATURES
}


# ==========================================================
# 5. RESULTS STORAGE
# ==========================================================

results = pd.DataFrame(
    columns=[
        "Period",
        "Target",
        "Horizon",
        "Model",
        "MAE",
        "RMSE",
        "MAPE",
        "R2"
    ]
)

# ==========================================================
# MAPE FUNCTION
# ==========================================================

def calculate_mape(
    y_true,
    y_pred
):

    y_true = np.where(
        y_true == 0,
        1e-8,
        y_true
    )

    mape = np.mean(
        np.abs(
            (y_true - y_pred)
            / y_true
        )
    ) * 100

    return mape

# ==========================================================
# 6. EVALUATION FUNCTIONS
# ==========================================================

def evaluate_model(
    y_true,
    y_pred
):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    mape = calculate_mape(
        y_true,
        y_pred
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    return (
        mae,
        rmse,
        mape,
        r2
    )

####################################################

def add_result(
    results,
    period,
    target,
    horizon,
    model,
    mae,
    rmse,
    mape,
    r2
):

    results.loc[len(results)] = {

        "Period": period,
        "Target": target,
        "Horizon": horizon,
        "Model": model,

        "MAE": mae,
        "RMSE": rmse,
        "MAPE": mape
        "R2": r2

    }

    return results


# ==========================================================
# 7. DATA SPLITTING FUNCTIONS
# ==========================================================
def get_period_split(df, period_name):

    period = PERIODS[period_name]

    # Filter period
    df_period = df[
        df["datetime"] >= period["start"]
    ].copy()

    # Train
    train_df = df_period[
        df_period["datetime"] <= period["train_end"]
    ].copy()

    # Validation
    valid_df = df_period[
        (df_period["datetime"] > period["train_end"]) &
        (df_period["datetime"] <= period["valid_end"])
    ].copy()

    # Test
    test_df = df_period[
        df_period["datetime"] > period["valid_end"]
    ].copy()

    return train_df, valid_df, test_df

####################################################################

def get_model_data(
    train_df,
    valid_df,
    test_df,
    target_name,
    horizon
):

    features = FEATURES[target_name]

    target_col = f"target_h{horizon}"

    X_train = train_df[features]
    y_train = train_df[target_col]

    X_valid = valid_df[features]
    y_valid = valid_df[target_col]

    X_test = test_df[features]
    y_test = test_df[target_col]

    return (
        X_train,
        y_train,
        X_valid,
        y_valid,
        X_test,
        y_test
    )


# ==========================================================
# 8. MODEL FUNCTIONS
# ==========================================================


def run_persistence(
    valid_df,
    target_name,
    horizon
):

    target_col = f"target_h{horizon}"

    base_target = TARGETS[target_name]

    lag_col = f"{base_target}_lag_{horizon}"

    y_true = valid_df[target_col]

    y_pred = valid_df[lag_col]

    mae, rmse, mape, r2 = evaluate_model(
        y_true,
        y_pred
    )

    return mae, rmse, mape, r2

#################################################################

def run_random_forest(
    X_train,
    y_train,
    X_valid,
    y_valid
):

    model = RandomForestRegressor(
        n_estimators=200,
        max_depth=15,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train,
        y_train
    )

    y_pred = model.predict(
        X_valid
    )

    mae, rmse, mape, r2 = evaluate_model(
        y_valid,
        y_pred
    )

    return model, mae, rmse, mape, r2

#########################################################################################
def run_xgboost(
    X_train,
    y_train,
    X_valid,
    y_valid
):

    model = XGBRegressor(
        n_estimators=500,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train,
        y_train
    )

    y_pred = model.predict(
        X_valid
    )

    mae, rmse, mape, r2 = evaluate_model(
        y_valid,
        y_pred
    )

    return model, mae, rmse, mape, r2
    
#############################################################


def run_lightgbm(
    X_train,
    y_train,
    X_valid,
    y_valid
):

    model = LGBMRegressor(
        n_estimators=500,
        learning_rate=0.05,
        max_depth=6,
        num_leaves=31,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train,
        y_train
    )

    y_pred = model.predict(
        X_valid
    )

    mae, rmse, mape, r2 = evaluate_model(
        y_valid,
        y_pred
    )

    return model, mae, rmse, mape, r2


# ==========================================================
# 9. MODEL DISPATCHER
# ==========================================================


def run_model(model_name, 
              X_train=None, y_train=None, 
              X_valid=None, y_valid=None, 
              valid_df=None, target_name=None, horizon=None ):

    # =====================================
    # Persistence
    # =====================================
    if model_name == "Persistence":

        mae, rmse, mape, r2 = run_persistence(
            valid_df=valid_df,
            target_name=target_name,
            horizon=horizon
        )

        return {
            "model": None,
            "mae": mae,
            "rmse": rmse,
            "mape": mape
            "r2": r2
        }

    # =====================================
    # Random Forest
    # =====================================
    elif model_name == "RandomForest":

        model, mae, rmse, mape, r2 = run_random_forest(
            X_train,
            y_train,
            X_valid,
            y_valid
        )

        return {
            "model": model,
            "mae": mae,
            "rmse": rmse,
            "mape": mape,
            "r2": r2
        }

    # =====================================
    # XGBoost
    # =====================================
    elif model_name == "XGBoost":

        model, mae, rmse, mape, r2 = run_xgboost(
            X_train,
            y_train,
            X_valid,
            y_valid
        )

        return {
            "model": model,
            "mae": mae,
            "rmse": rmse,
            "mape": mape,
            "r2": r2
        }

    # =====================================
    # LightGBM
    # =====================================
    elif model_name == "LightGBM":

        model, mae, rmse, mape, r2 = run_lightgbm(
            X_train,
            y_train,
            X_valid,
            y_valid
        )

        return {
            "model": model,
            "mae": mae,
            "rmse": rmse,
            "mape": mape,
            "r2": r2
        }

    # =====================================
    # LSTM
    # =====================================
    elif model_name == "LSTM":

        model, mae, rmse, mape, r2 = run_lstm(
            X_train,
            y_train,
            X_valid,
            y_valid
        )

        return {
            "model": model,
            "mae": mae,
            "rmse": rmse,
            "mape": mape,
            "r2": r2
        }

    # =====================================
    # Unknown Model
    # =====================================
    else:

        raise ValueError(
            f"Unknown model: {model_name}"
        )




# ==========================================================
# 10. EXPERIMENT RUNNER
# ==========================================================
def run_experiment(
    dataset,
    period_name,
    target_name,
    horizon,
    model_name,
    results
):

    # Split data
    train_df, valid_df, test_df = get_period_split(
        dataset,
        period_name
    )

    # Features + Target
    (
        X_train,
        y_train,
        X_valid,
        y_valid,
        X_test,
        y_test
    ) = get_model_data(
        train_df,
        valid_df,
        test_df,
        target_name,
        horizon
    )

    # Run model
    result = run_model(
        model_name=model_name,

        X_train=X_train,
        y_train=y_train,

        X_valid=X_valid,
        y_valid=y_valid,

        valid_df=valid_df,

        target_name=target_name,
        horizon=horizon
    )

    # Save result
    results = add_result(
        results,
        period_name,
        target_name,
        horizon,
        model_name,
        result["mae"],
        result["rmse"],
        result["mape"],
        result["r2"]
    )

    return results


# ==========================================================
# 11. MASTER EXPERIMENT LOOP
# ==========================================================
TOTAL_RUNS = (
    len(PERIODS)
    * len(TARGETS)
    * len(HORIZONS)
    * len(MODEL_NAMES)
)
print("Total Experiments:", TOTAL_RUNS)
counter = 1

for period_name in PERIODS:

    for target_name in TARGETS:
        dataset = DATASETS[target_name]

        for horizon in HORIZONS:

            for model_name in MODEL_NAMES:

                print(
                    f"[{counter}/{TOTAL_RUNS}] "
                    f"{period_name} | "
                    f"{target_name} | "
                    f"H{horizon} | "
                    f"{model_name}"
                )

                counter += 1

                try:
                    results = run_experiment(
                        dataset=dataset,
                        period_name=period_name,
                        target_name=target_name,
                        horizon=horizon,
                        model_name=model_name,
                        results=results
                    )
                except Exception as e:
                    
                    print(
                        f"FAILED: "
                        f"{period_name} | "
                        f"{target_name} | "
                        f"H{horizon} | "
                        f"{model_name}"
                    )
                
                    print(e)

# ==========================================================
# 12. SAVE RESULTS
# ==========================================================

os.makedirs(
    "results",
    exist_ok=True
)

results.to_csv(
    "results/validation_results.csv",
    index=False
)

print(results.shape)
results.head()


################################################
end_time = time.time()

print(
    f"Total runtime: "
    f"{(end_time-start_time)/60:.2f} minutes"
)
```

```python
results = pd.read_csv( "results/validation_results.csv" )
# ==========================================================
# MODEL RANKING TABLE
# ==========================================================

model_ranking = (
    results
    .groupby("Model")
    .agg({
        "MAE": "mean",
        "RMSE": "mean",
        "R2": "mean"
    })
    .sort_values("RMSE")
    .reset_index()
)

print(model_ranking)

model_ranking.to_csv(
    "results/model_ranking_table.csv",
    index=False
)

# ==========================================================
# HORIZON COMPARISON TABLE
# ==========================================================

horizon_table = (
    results
    .groupby("Horizon")
    .agg({
        "MAE": "mean",
        "RMSE": "mean",
        "R2": "mean"
    })
    .reset_index()
    .sort_values("Horizon")
)

print(horizon_table)

horizon_table.to_csv(
    "results/horizon_comparison_table.csv",
    index=False
)

# ==========================================================
# PERIOD COMPARISON TABLE
# ==========================================================

period_table = (
    results
    .groupby("Period")
    .agg({
        "MAE": "mean",
        "RMSE": "mean",
        "R2": "mean"
    })
    .sort_values("RMSE")
    .reset_index()
)

print(period_table)

period_table.to_csv(
    "results/period_comparison_table.csv",
    index=False
)


# ==========================================================
# BEST CONFIGURATION TABLE
# ==========================================================

best_configs = (
    results
    .loc[
        results.groupby(
            ["Target", "Horizon"]
        )["RMSE"].idxmin()
    ]
    .sort_values(
        ["Target", "Horizon"]
    )
)

print(best_configs)

best_configs.to_csv(
    "results/best_configuration_table.csv",
    index=False
)
```
